```python
client = genai.Client(api_key="")
```

```python
MODEL = "models/gemini-3.5-flash"
```

```python
def travel_agent(user_input):

    prompt = f"""
You are a professional travel agent for Pakistan tourism.

Provide a well-structured travel plan with clear headings:

1. Destination Overview
2. Budget Travel Options (Bus, Train, Flight with approx cost in PKR)
3. Places to Visit (specific locations)
4. Food Suggestions (local dishes)
5. Money-Saving Tips

Make response detailed but easy for students.

User request: {user_input}

Travel Plan:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text



print("Gemini Travel Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye ")
        break

    try:
        reply = travel_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```

```python
!pip install -U google-genai
from google import genai

client = genai.Client(api_key="")
MODEL = "gemini-3.5-flash"

def agriculture_agent(user_input):

    prompt = f"""
You are a helpful Smart Agriculture Assistant designed for farmers and agriculture
students in Pakistan.

Answer the user's question in easy to understand way. Explain things as an
experienced agriculture advisor would explain them to a farmer.
 Consider factors such as location, crop season, soil condition, water availability, farm size, and budget
whenever the user provides them.

Ask the answer with these sections when they are relevant:

1. What is the crop/problem?
2. Recommended crop or practical solution
3. Irrigation and water management
4. Fertilizer and nutrient management
5. Pest and disease control
6. Weather and seasonal advice
7. Useful smart farming technologies
8. Practical ways to save money and improve productivity


Give practical steps that a farmer could actually follow, explain agricultural terms briefly when they are
important.

If the user's question is simple, give a focused answer instead of forcing all
10 sections into the response. If the question requires a complete farming plan,
then cover the relevant sections in detail.

User's question:
{user_input}

Now provide a practical and natural Smart Agriculture response suitable for
Pakistan.
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("Smart Agriculture Assistant is live! (Type 'exit' to stop)\n")

while True:
    user_msg = input("You: ")

    if user_msg.strip().lower() == "exit":
        print("Bye, good luck with your farming!")
        break

    if not user_msg.strip():
        continue

    try:
        reply = agriculture_agent(user_msg)
        print("\nAssistant:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
