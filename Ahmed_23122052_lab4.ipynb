```python
!pip install -U google-genai
from google import genai

client = genai.Client(api_key="")

MODEL = "models/gemini-3.5-flash"

cs_info = {
    "department": "Computer Science (FUUAST)",
    "timings": "9:00 AM to 3:40 PM (Monday to Friday)",
    "lecture_duration": "50 minutes per class",
    "break_time": "1:10 PM to 2:00 PM",
    "rooms_and_labs": "Lab 1 to 4, Room 1, 2, 3, 11, 12, 13, Hall, and Library",
    "semesters_courses": {
        "BS 2nd Semester": "OOP, Digital Logic Design (DLD), Pre-Calculus, Pak Studies, Arts and Humanities, Quantitative Reasoning, Fehm-e-Quran",
        "BS 4th Semester": "Database (DBMS), Theory of Automata, Linear Algebra, Design and Analysis of Algorithms, Accounting",
        "BS 6th Semester": "Artificial Intelligence (AI), Natural Language Processing (NLP), Compiler Construction, Numerical Computing, Professional Practices",
        "BS 7th Semester": "Data Science, Parallel and Distributed Computing, Cloud Computing, Entrepreneurship",
        "BS 8th Semester": "Human Computer Interaction (HCI), Wireless Networks"
    }
}

def ask_agent(question):
    prompt = f"""
You are a friendly student help-desk assistant for the FUUAST Computer Science (CS) Department.

Here is the basic information you know:
- Department: {cs_info['department']}
- Timings: {cs_info['timings']}
- Lecture Duration: {cs_info['lecture_duration']}
- Break Time: {cs_info['break_time']}
- Rooms and Labs: {cs_info['rooms_and_labs']}
- Courses Offered: {cs_info['semesters_courses']}

Instructions:
1. Only answer questions related to the CS department, semester courses, and timetable.
2. Keep your tone casual, friendly, and helpful.
3. Keep your answers short, simple, and easy to understand.

Student question: "{question}"

Helpful response:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("FUUAST CS Help Desk is live! (Type 'exit' to stop)\n")

while True:
    user_msg = input("You: ")

    if user_msg.strip().lower() == "exit":
        print("Bye, good luck!")
        break

    try:
        reply = ask_agent(user_msg)
        print("\nAssistant:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
